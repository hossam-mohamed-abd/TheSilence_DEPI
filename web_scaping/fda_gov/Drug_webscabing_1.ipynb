```python
import requests
import os

headers = {
    "User-Agent": "Mozilla/5.0"
}

years_data = {
    2023: {
        "q1": 30,
        "q2": 29,
        "q3": 29,
        "q4": 29
    },

    2024: {
        "q1": 29,
        "q2": 29,
        "q3": 29,
        "q4": 29
    },
      2025: {
        "q1": 28,
        "q2": 28,
        "q3": 34,
        "q4": 30
    }
}

base_root = "https://download.open.fda.gov/drug/event/"

for year, quarters in years_data.items():

    # فولدر السنة
    year_folder = f"openfda/{year}"
    os.makedirs(year_folder, exist_ok=True)

    for quarter, total_parts in quarters.items():

        print(f"\n=== Downloading {year} {quarter.upper()} ===")

        # فولدر الـ quarter
        quarter_folder = f"{year_folder}/{quarter}"
        os.makedirs(quarter_folder, exist_ok=True)

        base_url = f"{base_root}{year}{quarter}/"

        for i in range(1, total_parts + 1):

            file_num = str(i).zfill(4)
            total_str = str(total_parts).zfill(4)

            filename = f"drug-event-{file_num}-of-{total_str}.json.zip"

            file_url = base_url + filename

            print("Downloading:", filename)

            try:

                response = requests.get(file_url, headers=headers)

                if response.status_code == 200:

                    with open(f"{quarter_folder}/{filename}", "wb") as f:
                        f.write(response.content)

                else:
                    print("Failed:", response.status_code)

            except Exception as e:
                print("Error:", e)

print("\nAll downloads completed")
```
